```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# 1. Load Dataset
df = pd.read_csv('../data/raw_sales_data.csv')

# 2. Data Cleaning Routine
# Remove duplicate records
df.drop_duplicates(inplace=True)

# Handle missing numerical values with median imputation
if 'revenue' in df.columns:
    df['revenue'] = df['revenue'].fillna(df['revenue'].median())

# Format date column
df['order_date'] = pd.to_datetime(df['order_date'])

# Outlier Detection via IQR for Revenue
Q1 = df['revenue'].quantile(0.25)
Q3 = df['revenue'].quantile(0.75)
IQR = Q3 - Q1
df_cleaned = df[(df['revenue'] >= (Q1 - 1.5 * IQR)) & (df['revenue'] <= (Q3 + 1.5 * IQR))]

# Export Cleaned Data
df_cleaned.to_csv('../data/cleaned_sales_data.csv', index=False)

# 3. KPI Calculation & Summaries
print("=== Key Business Performance Indicators ===")
print(f"Total Revenue: ${df_cleaned['revenue'].sum():,.2f}")
print(f"Average Order Value: ${df_cleaned['revenue'].mean():,.2f}")
print(f"Median Order Value: ${df_cleaned['revenue'].median():,.2f}")
print(f"Revenue Standard Deviation: ${df_cleaned['revenue'].std():,.2f}")

# 4. Data Visualization
plt.figure(figsize=(10, 5))
sns.histplot(df_cleaned['revenue'], kde=True, color='skyblue')
plt.title('Distribution of Cleaned Transaction Revenue')
plt.xlabel('Revenue ($)')
plt.ylabel('Frequency')
plt.savefig('../reports/revenue_distribution.png')
plt.show()
```
